```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
df = pd.read_csv('../data/raw/202608_CPGF.csv', sep = ';', encoding= 'latin-1')
df.info()
```

```python
df['VALOR TRANSAÇÃO'] = df['VALOR TRANSAÇÃO'].str.replace(',', '.').astype(float)
```

```python
df['DATA TRANSAÇÃO'] = pd.to_datetime(df['DATA TRANSAÇÃO'], format='%d/%m/%Y')
```

```python
df.isna().sum()
```

```python
df['CPF PORTADOR'] = df['CPF PORTADOR'].fillna('Sigiloso')
```

```python
engine = create_engine('postgresql://user:password@localhost:0000/db_transparencia')
```

```python
df.to_sql(name = 'despesas_cpgf', con = engine, if_exists = 'replace', index = False)
```
